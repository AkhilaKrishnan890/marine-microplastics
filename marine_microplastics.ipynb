## Marine Microplastics Pollution Analysis

Exploratory Data Analysis (EDA) is the process of examining, cleaning, and visualizing a dataset to understand its structure, patterns, distributions, and relationships before applying statistical or machine learning techniques.

In this project, EDA is performed on a Marine Microplastics Pollution dataset to understand different aspects of marine pollution, such as ocean and region distribution, sampling methods, microplastics measurements, density classifications, geographical locations, and observation dates.

Steps for an Exploratory Data Analysis (EDA) Project

1. Data Collection

* Collect the Marine Microplastics dataset from the given source.

* Load the dataset using Python and Pandas.

* Check the number of rows and columns.

2. Data Cleaning

* Check for missing values and duplicate records.

* Correct incorrect data types and date formats.

* Check inconsistent measurement units and unusual values.

3. Data Understanding

* Examine the structure and understand the meaning of each column.

* Identify numerical and categorical variables.

* Use summary statistics and check unique values.

4. Univariate Analysis

* Analyze one variable at a time.

* Study the distribution of oceans, regions, sampling methods, and density classes.

* Use bar charts, histograms, and boxplots.

5. Bivariate Analysis

* Explore relationships between two variables.

* Compare microplastics measurements across oceans and regions.

* Analyze sampling methods, density classes, and measurements using suitable charts.

6. Multivariate Analysis

* Analyze three or more variables together.

* Compare oceans, regions, sampling methods, and density classes.

* Use groupby, pivot tables, and heatmaps to identify relationships.

7. Feature Engineering

* Extract year and month from the date column.

* Create meaningful measurement or density categories where appropriate.

* Add useful features to support further analysis.

8. Outlier Detection

* Identify unusual values in numerical columns.

* Use boxplots and the Interquartile Range (IQR) method to detect potential outliers.

* Check whether extreme values are valid observations or data-quality issues.

9. Trend and Pattern Analysis

* Analyze yearly patterns in observations, if date information is available.

* Identify oceans and regions with the highest number of recorded observations.

* Compare measurements and geographical patterns across different locations.

10. Documentation & Reporting

* Summarize the important findings from the analysis.

* Present insights using tables, charts, and graphs.

* Explain the major findings, limitations, and conclusions.

### Feature Engineering

#### 34. Can we extract the year, month and month name from the Date column?

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()

df[["Date", "Year", "Month", "Month_Name"]].head()
```

#### 35. How many observations were recorded in each year?

Insight: This shows how observations are distributed across years. More observations in a particular year do not necessarily mean pollution increased during that year.

### Outlier Detection

#### 36. Are there outliers in the Measurement column using the IQR method?

```python
Q1 = df["Measurement"].quantile(0.25)
Q3 = df["Measurement"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Measurement"] < lower_bound) |
    (df["Measurement"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))

outliers.head()
```

#### 37. Can we visualise the potential outliers?

```python
plt.figure(figsize=(9, 4))

sns.boxplot(data=df, x="Measurement")

plt.title("Outlier Detection in Microplastic Measurements")
plt.xlabel("Measurement")
plt.show()
```

Insight: The boxplot shows the central spread of measurements and potential outliers beyond the whiskers. An outlier is not automatically an incorrect value, so do not remove it without investigation.

### Trend and Pattern Analysis

#### 38. How do average measurements change over time?

```python
yearly_measurement = (
    df.dropna(subset=["Year", "Measurement"])
      .groupby("Year")["Measurement"]
      .mean()
      .sort_index()
)

print(yearly_measurement)

yearly_measurement.plot(
    kind="line",
    marker="o",
    figsize=(10, 5)
)

plt.title("Yearly Trend in Average Microplastic Measurements")
plt.xlabel("Year")
plt.ylabel("Average Measurement")
plt.grid(True)
plt.show()
```

Insight: This line chart shows whether average recorded measurements increase, decrease or fluctuate over time. Interpret the trend carefully because changes in sampling coverage, methods or units can affect the results.

#### 39. Which months have the highest average measurements?

```python
monthly_measurement = (
    df.dropna(subset=["Month", "Measurement"])
      .groupby("Month")["Measurement"]
      .mean()
      .reindex(range(1, 13))
)

print(monthly_measurement)

monthly_measurement.plot(
    kind="bar",
    figsize=(10, 5),
    color=["green"]
)

plt.title("Average Microplastic Measurements by Month")
plt.xlabel("Month")
plt.ylabel("Average Measurement")

plt.xticks(
    ticks=range(12),
    labels=[
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],
    rotation=45
)

plt.tight_layout()
plt.show()
```

Insight: This compares average measurements by calendar month. A seasonal pattern should only be suggested if the dataset contains enough observations across comparable periods and years.

#### 40. Average Measurement by Ocean

```python

ocean_avg = df.groupby("Oceans")["Measurement"].mean().sort_values(ascending=False)

print("Average Measurement by Ocean:")
print(ocean_avg)

plt.figure(figsize=(10, 6))

ocean_avg.plot(
    kind="bar",
    color="#6A0572",
    edgecolor="black"
)

plt.title("Average Microplastic Measurement by Ocean")
plt.xlabel("Ocean")
plt.ylabel("Average Measurement")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Limitations

The dataset may contain missing values, unusual observations, or inconsistencies that can affect the accuracy of the analysis.

The dataset covers specific sampling locations, marine environments, and observation periods. Therefore, the findings may not represent all oceans, seas, or coastal ecosystems.

Relationships identified through correlation analysis do not necessarily indicate cause-and-effect relationships between environmental factors and microplastic pollution.

Factors such as ocean currents, seasonal changes, human activities, wastewater discharge, and coastal industrialization may influence microplastic distribution but may not be fully represented in the dataset.

Differences in sampling methods, laboratory procedures, particle-detection techniques, and reporting units may affect comparisons between observations.

Outliers may represent genuine cases of high microplastic accumulation rather than data errors, so they should not be removed without proper investigation.

Missing information about sampling locations or environmental conditions may limit detailed geographical and environmental comparisons.

The analysis mainly focuses on exploratory techniques and descriptive statistics. It does not independently establish the long-term effects of microplastics on marine organisms or predict future pollution levels.

Conclusions about pollution trends over time depend on the availability, consistency, and quality of the recorded dates and observations.
